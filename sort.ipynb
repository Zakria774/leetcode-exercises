```python
def bubble_sort(nums):
    for i in range(len(nums)-1):
        for j in range(len(nums)-1):
            if nums[j] > nums[j+1]:
                nums[j], nums[j+1] = nums[j+1], nums[j]
```

```python
data=[1,4,3,2,9]
bubble_sort(data)
print(data)
```

```python
def insertion_sort(nums):
    nums=list(nums)
    for i in range(1,len(nums)):
        value=nums[i]
        while i>0 and nums[i-1]>value:
            nums[i],nums[i-1]=nums[i-1],nums[i]
        i-=1
    return nums

```

```python
insertion_sort(data)
```

```python
def merge(num1, num2):
    merged_list=[]
    i, j=0,0
    while i<len(num1) and j<len(num2):
        if num1[i]<num2[j]:
            merged_list.append(num1[i])
            i+=1
        else:
            merged_list.append(num2[j])
            j+=1
        last_i=num1[i:]
        last_j=num2[j:]

        return merged_list+last_i+last_j
```

```python

def merge_sort(nums):
    if len(nums) <=1:
        return nums
    mid=len(nums)//2

    left_num=nums[:mid]
    right_num=nums[mid:]

    left_sorted, right_sorted=merge_sort(left_num), merge_sort(right_num)
    return merge(left_sorted, right_sorted)
```

```python
merge_sort(data)
```

```python
def partition(nums, start=0, end=None):
    if end==None:
        end=len(nums)-1
    l, r=start,end-1

    while r>l:
        if nums[l]<=nums[end]:
            l+=1
        elif nums[r]>=nums[end]:
            r-=1
        else:
            nums[l],nums[r]=nums[r],nums[l]
    if nums[l]>nums[end]:
        nums[end],nums[l]=nums[l],nums[end]
        return l
    else:
        return end
```

```python
def quick_sort(nums, start=0, end=None):
    if end==None:
        nums=list(nums)
        end=len(nums)-1
    if start<end:
        pivot=partition(nums, start, end)
        quick_sort(nums, start, pivot-1)
        quick_sort(nums, pivot+1, end)
    return nums
```

```python
quick_sort(data)
```

```python
def quicksort(nums):
    if len(nums)<=1:
        return nums
    else:
        pivot=nums.pop()
    items_lower=[]
    items_greater=[]
    for item in nums:
        if item<pivot:
            items_lower.append(item)
        else:
            items_greater.append(item)
    return quicksort(items_lower)+[pivot]+items_greater+quicksort(items_greater)
```

```python
quicksort(data)
```
